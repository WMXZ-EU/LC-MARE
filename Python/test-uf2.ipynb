```python
import serial
import serial.tools.list_ports

def getComPort():
    s=serial.tools.list_ports.comports(True)
    for ii in range(len(s)):
        if (s[ii].vid==0x239a) & (s[ii].pid==0x815d): # adafruit adalogger rp2040
            return s[ii].device
    return None

comPort=getComPort()
print(comPort)

program="adalogger_V3"
path="../adalogger_V3/build/rp2040.rp2040.adafruit_feather_adalogger/"

cmd="python -I uf2conv.py -s "+comPort+" -f rp2040 -D "+path+program+".ino.uf2 "
import subprocess

# Execute a shell command
result = subprocess.run(cmd, shell=True, capture_output=True, text=True)

# Print the output
print("Output:\n", result.stdout)
print("Error:", result.stderr)
print("Return Code:", result.returncode)
```
