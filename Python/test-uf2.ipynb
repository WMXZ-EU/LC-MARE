```python
import serial
import serial.tools.list_ports

# find first device on com-port
[vid,pid]= [0x239a,0x815d]  # adafruit adalogger rp2040

def getComPort():
    s=serial.tools.list_ports.comports(True)
    for ii in range(len(s)):
        if (s[ii].vid==vid) & (s[ii].pid==pid): return s[ii].device
    return None

comPort=getComPort()
print(comPort)

# the following could be used if LC-MARE is downloaded via GitHub
program="adalogger_V3"
program_path="../"+program+"/build/rp2040.rp2040.adafruit_feather_adalogger/"+program+".ino.uf2"

cmd="python -I uf2conv.py -s "+comPort+" -f rp2040 -D "+program_path
import subprocess

# Execute a shell command
result = subprocess.run(cmd, shell=True, capture_output=True, text=True)

# Print the output
print("Output:\n", result.stdout)
print("Error:", result.stderr)
print("Return Code:", result.returncode)
```
